# Weather dataset analysis

Explore the bundled hourly Mumbai example. The small sample demonstrates the quality checks and charts; it is not large enough for operational weather forecasts.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

candidates = [Path("backend/datasets/sample_weather.csv"), Path("../backend/datasets/sample_weather.csv")]
dataset_path = next(path for path in candidates if path.exists())
weather = pd.read_csv(dataset_path, parse_dates=["date"])
weather.head()
```

```python
print("Rows:", len(weather))
display(weather.dtypes.rename("dtype").to_frame())
print("Missing values:")
display(weather.isna().sum().rename("missing").to_frame())
print("Duplicate timestamps:", weather.duplicated(["date", "latitude", "longitude"]).sum())
display(weather.describe().round(2))
```

```python
valid_ranges = {"temperature": (-90, 65), "humidity": (0, 100), "rainfall": (0, 1000), "wind_speed": (0, 500), "pressure": (800, 1200)}
outliers = {column: int((~weather[column].between(low, high)).sum()) for column, (low, high) in valid_ranges.items()}
pd.Series(outliers, name="outlier_count").to_frame()
```

```python
weather = weather.sort_values("date")
fig, axes = plt.subplots(4, 1, figsize=(12, 10), sharex=True)
for axis, column, color in zip(axes, ["temperature", "humidity", "rainfall", "wind_speed"], ["#e8a23a", "#5278e5", "#48a4ce", "#35a484"]):
    axis.plot(weather["date"], weather[column], color=color, linewidth=1.8)
    axis.set_ylabel(column.replace("_", " "))
    axis.grid(alpha=0.2)
fig.suptitle("Mumbai hourly weather observations")
fig.tight_layout()
```
